```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#import seaborn as sns

# 1. Load the results from your Random Forest run
# (Assuming performance_df is already in your memory from the previous loop)
df_results = pd.read_csv('results/plsr_performance_results_combinations.csv') 
#df_results = performance_df.copy()

# 2. Find the configuration with the LOWEST MAE for each Feature
# We use idxmin() to find the index of the minimum Mean Absolute Error
best_rf_configs = df_results.loc[df_results.groupby('Feature')['MAE'].idxmin()]

# 3. Clean up the table for a clear summary
summary_table = best_rf_configs[['Feature', 'Config', 'MAE', 'R2', 'Best_Components']].reset_index(drop=True)

# 4. Save this 'Winner List' 
# This file becomes your blueprint for the Stacking phase
summary_table.to_csv('plsr_winning_configs_blueprint.csv', index=False)

print("--- Winning RF Configurations per Feature ---")
print(summary_table)
```
