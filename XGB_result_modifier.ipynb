```python
import pandas as pd

# 1. Define your filename (change this to your actual XGBoost CSV filename)
file_name = 'results\model_configs\XGBoost_final_configs(1).csv' 
save_file = "results\model_configs\XGBoost_final_configs.csv"
# If you are doing this on the blueprint file, use 'xgb_winning_configs_blueprint.csv'

# Load the dataframe
df = pd.read_csv(file_name)

# Parameter columns to combine
param_cols = ['n_estimators', 'learning_rate', 'max_depth', 'subsample', 'colsample_bytree']
import pandas as pd

# Assuming your dataframe is called 'df'
# First, make sure the column exists and is set to hold objects (like dicts)
if 'Best_Params' not in df.columns:
    df['Best_Params'] = None 
    df['Best_Params'] = df['Best_Params'].astype(object)

# Iterate through the dataframe using the index
for index in df.index:
    
    # 1. Grab whatever data you need from the current row
    # example_value = df.at[index, 'some_other_column']
    new_params = {}
    for param in param_cols:
        new_params[param] = df.at[index, param].item()

    # 3. Assign the dictionary directly to the cell using .at
    df.at[index, 'Best_Params'] = new_params
    print(new_params)


print(df.head())

# 6. Save it back (overwriting the file or saving as a new one)
df.to_csv(save_file, index=False)

print(f"Successfully added 'Best_Params' and updated {save_file}!")
```
